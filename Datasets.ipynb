# Heart Disease UCI Dataset

## Description

The Heart Disease UCI Dataset is a multivariate dataset containing health and clinical information about patients. It includes 14 commonly used attributes related to heart disease. The dataset can be used to explore relationships between patient characteristics, clinical measurements, and the presence of heart disease.

## Columns / Data Dictionary

| Column | Description |
|---|---|
| **id** | Unique identifier for each patient. |
| **age** | Age of the patient in years. |
| **origin** | Place where the study was conducted. |
| **sex** | Sex of the patient (Male/Female). |
| **cp** | Chest pain type: 1 = typical angina, 2 = atypical angina, 3 = non-anginal, 4 = asymptomatic. |
| **trestbps** | Resting blood pressure in mm Hg on admission to the hospital. |
| **chol** | Serum cholesterol level in mg/dl. |
| **fbs** | Indicates whether fasting blood sugar is greater than 120 mg/dl. |
| **restecg** | Resting electrocardiographic results: normal, ST-T wave abnormality, or left ventricular hypertrophy. |
| **thalach** | Maximum heart rate achieved. |
| **exang** | Exercise-induced angina (True/False). |
| **oldpeak** | ST depression induced by exercise relative to rest. |
| **slope** | Slope of the peak exercise ST segment. |
| **ca** | Number of major vessels (0–3) colored by fluoroscopy. |
| **thal** | Thalassemia result: normal, fixed defect, or reversible defect. |
| **num** | Predicted attribute indicating the presence or severity of heart disease. |

## Purpose of the Dataset

The dataset can be used to investigate relationships between clinical characteristics and heart disease. It can also be used for data visualization, statistical analysis, feature selection, and machine learning models that predict heart disease.
